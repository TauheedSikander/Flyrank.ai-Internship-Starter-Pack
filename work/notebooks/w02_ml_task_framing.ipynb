## 3. Success metric

## Success Metric

The primary success metric for this project is **Precision@50**. This measures how many of the top 50 webpages recommended by the model are actually declining pages.

Precision@50 is appropriate because SEO teams usually review only a limited number of pages at a time. A higher Precision@50 means that more of the recommended pages genuinely need attention, making the recommendations more useful for prioritizing content updates.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML Beats a Fixed Rule

A simple rule, such as refreshing pages that are old and receive many impressions, may work in some situations but cannot capture the complex relationships between multiple factors.

Machine learning can analyze several signals simultaneously, including impressions, click-through rate (CTR), average position, content age, word count, and days since the last update. By learning patterns from historical data, the model can produce a more accurate ranking of webpages than a single hand-written rule. This helps SEO teams prioritize their work more effectively while reducing manual effort.

```python
features = [
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "avg_position",
    "ctr",
    "word_count"
]

print("Features used for ML:")
for feature in features:
    print("-", feature)
```
